```python
def mean(data):
  return np.sum(data)/len(data)
```

```python
def variance(data):
  m=mean(data)
  v=np.sum((data-m)**2)/len(data)
  return v
```

```python
import numpy as np

# -------------------------
# Dataset
# -------------------------

X = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([5,7,9,11,13]).reshape(-1,1)

# -------------------------
# Initialize parameters
# -------------------------

w = np.random.randn(1)

print(X.shape)
print(y.shape)
print(w.shape)

b = 100

# -------------------------
# Prediction function
# -------------------------

def predict(X, w, b):
    return X * w + b

# -------------------------
# Training loop
# -------------------------

lr = 0.01
epochs = 10000
n = len(X)

for i in range(epochs):

    y_pred = predict(X, w, b)

    error = y_pred - y

    if i==1:
      print(y_pred.shape)
      print(error.shape)

    dw = (2/n) * np.sum(X * error)
    db = (2/n) * np.sum(error)

    w = w - lr * dw
    b = b - lr * db

    if i % 1000 == 0:
        loss = np.mean(error**2)
        print("Epoch:", i, "Loss:", loss)

# -------------------------
# Final prediction
# -------------------------

print("Learned w:", w)
print("Learned b:", b)

print("Prediction for x=6:", predict(np.array([6]), w, b))
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([
    [1000,2],
    [1500,3],
    [2000,4]
])

y = np.array([300,450,600])

w = np.random.randn(2)
b = 0

lr = 1e-8
epochs=1000
n=len(X)

losses=[]

for i in range(epochs):

  Y= np.dot(X,w) + b

  error = Y - y

  loss = (1/n) * np.sum(error**2)

  losses.append(loss)

  dw = (2/n) * np.dot(X.T,error)
  db = (2/n) * np.sum(error)

  w = w - lr * dw
  b = b - lr * db

  if i%1000 == 0:
    print("run: ",i,"loss: ",loss)

print(w,b)
print(np.dot(np.array([2500,5]),w)+b)

plt.plot(losses)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training Loss Curve")
plt.show()
```
